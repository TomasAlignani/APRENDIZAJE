# Introducción a la computación científica con Python

## El rol de la computación en la ciencia

Tradicionalmente la ciencia se ha dividido en disciplinas teóricas y experimentales. Sin embargo, en las últimas décadas la computación ha emergido como una parte muy importante de la ciencia. La computación científica está íntimamente relacionada con la teoría pero tiene muchas características del trabajo experimental. Por este motivo se ve como la tercera rama de la ciencia. En muchos campos de la ciencia, el trabajo computacional es un complemento importante tanto para la rama experimental como para la teórica. Hoy en día la gran mayoría de artículos de investigación tanto teóricos como experimentales incluyen cálculos numéricos, simulaciones o modelado por ordenador.

Tanto en las ciencias teóricas como experimentales existen códigos de conducta bien establecidos sobre cómo se deben publicar los resultados y los métodos para que estén disponibles para otros científicos. Por ejemplo, en ciencias teóricas, las demostraciones, derivaciones y otros resultados se publican en detalle o se facilitan cuando se reclaman. Del mismo modo, en ciencias experimentales, los métodos utilizados y los resultados obtenidos así como los datos experimentales deben ser facilitados cuando se reclamen. Está considerado no científico retener los detalles cruciales en una demostración teórica o el método experimental puesto que podría impedir replicar y reproducir los resultados a otros científicos.

En ciencias de computación no existen todavía guías bien establecidas sobre cómo se debe manejar el código fuente o los datos generados. Por ejemplo, es relativamente raro que se provea a los lectores de artículos publicados el código fuente utilizado en las simulaciones. Además, es bastante común retener el código fuente de simulaciones y que esta retención no sea considerada una ventaja competitiva (por tanto innecesario de publicar).

Sin embargo, este problema comienza a atraer la atención de la comunidad científica por lo que numerosas editoriales y revistas ciéntificas de prestigio demandan más claridad en las ciencias de la computación. Algunas revistas de prestigio, como Science, incluso han empezado a pedir a los autores  el código fuente de las simulaciones utilizadas en los artículos cuando sean reclamados. 

Se está trabajando en la manera de facilitar el software científico como por ejemplo incluirlo como material complementario en los artículos científicos.

## Requisitos de la computación científica

La **réplica** y la **reproducibilidad** son dos de los pilares del método científico. Con respecto a los trabajos numéricos, el cumplimiento de estos conceptos tiene las siguientes implicaciones prácticas:

* **Réplica**: un autor de un artículo científico que involucre cálculos numéricos debería poder ejecutar de nuevo la simulación y replicar los resultados cuando sea necesario. Además, utilizando la información publicada en un artículo sobre los métodos, otros científicos también deberían de ser capaces de realizar los mismos cálculos y obtener los mismos resultados.

* **Reproducibilidad**: los resultados obtenidos a partir de simulaciones numéricas deberían ser reproducibles utilizando una implementación independiente (diferente) del método dada la información publicada en un artículo científico.

En resumen: un resultado científico sólido debería ser reproducible y replicable.

Para conseguir estos objetivos, es necesario:

* Conocer exactamente el código fuente (y su versión) que se utilizó para producir los datos y las figuras publicadas en artículos científicos.

* Almacenar la información acerca de la versión de códigos externos que fueron utilizados en la experimentación. Mantener en acceso al entorno que fue utilizado en la experimentación.

* Realizar copias de seguridad de los códigos viejos (y las notas) para referencias futuras. 

* Estar preparado para ofrecer información adicional sobre los métodos utilizados (incluso de los códigos) a un lector interesado que lo reclame (incluso años después de la publicación del artículo).

* Idealmente los códigos se deberían publicar online para facilitar el acceso a los mismos a otros lectores.

## ¿Qué es Python?

[Python](http://www.python.org/) es un lenguaje de programación de alto nivel, moderno, de propósito general y orientado a objetos.

Características generales de Python:

* **Lenguaje simple y limpio:** el código realizado en Python es fácil de leer e intuitivo, utiliza una sintaxis fácil de aprender y escala bien con el tamaño de los proyectos.
* **Lenguaje expresivo:** Suele requerir menos líneas de código que otros lenguajes de programación, lo que implica cometer menos errores y facilita su mantenimiento.

Detalles técnicos:

* **Tipado dinámico:** No es necesario definir el tipo de datos que va a almacenar una variable, el tipo de datos de los argumentos de las funciones ni el tipo de datos que devuelven las funciones.
* **Gestión automática de la memoria:** No es necesario de reservar y liberar la memoria de forma explícita. Por tanto, se evitan los errores de este tipo. 
* **Lenguaje interpretado e interactivo:** No es necesario compilar el código. El intérprete de Python lee y ejecuta el código de Python directamente.

Ventajas:

* La principal ventaja es la **facilidad en la programación** que conlleva una **minimización en el tiempo** necesario para desarrollar, debuggear (corregir los errores) y mantener el código.
* Es un lenguaje bien diseñado que **estimula muchas buenas prácticas de programación**:
 * Programación modular y orientada a objetos, buen sistema para crear paquetes de código y reutilización del mismo. Estos factores normalmente implican la realización de códigos más transparentes, mantenibles y libres de errores.
 * Generación de documentación integrada en el código.
* Una librería estándar muy grande y una gran colección de paquetes adicionales.
* Dispone de una **gran comunidad de usuarios** por lo que es relativamente fácil encontrar información.
* Es un **lenguaje multiparadigma**: permite utilizar diferentes paradigmas de programación como el imperativo, el orientado a objetos o el funcional.

Desventajas:

* Puesto que Python es un lenguaje de programación interpretado y tipado dinámicamente, la ejecución de código Python puede ser más lenta que códigos compilados y tipados estáticamente como los realizados con C o Java. 
* Python permite trabajar en diferentes entornos, dispone de muchos paquetes y documentación desperdigada por muchos sitios. Estos factores pueden dificultar la curva de aprendizaje inicial.

## ¿Qué hace a Python adecuado para la computación científica?

* Python tiene una posición muy fuerte en la computación científica: 
    * Tiene una **gran comunidad de usuarios** que implican una facilidad de encontrar ayuda y documentación.

* Dispone de un **gran ecosistema de librerías científicas**
    * numpy (Numerical Python): http://numpy.scipy.org - Python numérico
    * scipy (Scientific Python): http://www.scipy.org -  Python científico
    * matplotlib (graphics library): http://www.matplotlib.org - Librería para realizar gráficos

* Gran **rendimiento** debido a su integración con códigos altamente optimizados escritos en C y Fortran:
    * blas, atlas blas, lapack, arpack, Intel MKL, ...

* Buen soporte para 
    * Procesamiento paralelo con procesos e hilos
    * Comunicación inter-procesos (MPI)
    * Computación GPU (OpenCL y CUDA)

* Disponibilidad inmediata y adecuada para el uso de **clústeres de computación** de alto rendimiento. 

* **Es libre** por lo que no existen costes por licencia y alivian los presupuestos de los proyectos.

### Entornos para ejecutar código Python

Python no solo es un lenguaje de programación sino que también se refiere a la implementación estándar del intérprete (técnicamente conocido como [CPython](http://en.wikipedia.org/wiki/CPython)) que es el encargado de ejecutar código Python en un ordenador.

Existen varios entornos diferentes a través de los cuales se puede utilizar el intérprete de Python. Cada entorno tiene sus ventajas y es adecuado para diferentes flujos de trabajo. Una fortaleza de Python es su versatilidad ya que puede ser utilizado de diferentes formas. Sin embargo, esto puede ser confuso para principiantes por lo que comenzamos con una breve descripción de los entornos de Python que son útiles para la computación científica.

#### Consola interactiva

Hay muchas maneras de usar el lenguaje Python. Se ha mencionado que es un lenguaje **interpretado** e **interactivo**. Si ejecutamos la consola (En windows `cmd.exe`) y luego `python`, se abrirá la consola interactiva

![](images/consolaPython.png)

<div class="alert alert-info">El símbolo >>> en la última línea indica que usted se encuentra en la sesión del intérprete interactiva de Python, también llamada “Python shell”. Este es diferente del símbolo de sistema normal del terminal!</div>

En la consola interactiva se pueden escribir sentencias o pequeños bloques de código que son ejecutados inmediatamente. Por ejemplo, en la imagen anterior se puede ver el código correspondiente a un programa que muestra el texto `Hola mundo` por pantalla. 

##### Programas

También podemos usar Python para hacer programas o scripts. Esto es, escribir nuestro código en un archivo con extensión .py y ejecutarlo con el intérprete de python. Por ejemplo, se puede escribir la instrucción mostrada en la consola interactiva (`print('Hola mundo')`) en un script de Python y llamarlo `hola.py`. A continuación, en la consola interactiva podemos ejecutar el script y el resultado será el mismo que el mostrado anteriormente.

![](images/consolaPythonScript.png)

<div class="alert alert-info">Para ejecutar scripts de Python, asegúrese de estar en la línea de comandos del sistema, el cual tendrá el símbolo > al final de la ruta, no el símbolo del intérprete de Python (que en cambio tiene >>>)!</div>

Sin embargo, este entorno es **limitado** por lo que no es la forma habitual de trabajar.

Para salir de la consola interactiva de Python se debe presiona `Control+z`.

#### IPython

IPython es un shell interactivo que aborda las limitaciones del intérprete estándar de Python y que es un modo de trabajo habitual. Ofrece una línea de comandos interactiva al intérprete de Python mucho más amigable que la consola interactiva.  Si ejecutamos la consola (`cmd.exe`) y luego `ipython`, se abrirá la consola correspondiente a `ipython`:

![](images/consolaIPython.png)

Algunas de las características útiles de IPython son:

* Historial de comandos, que puede ser consultado con las teclas del teclado con flechas en direcciones ascendente y descendente.
* Auto-completado utilizando la tecla de la tabulación (`<TAB>`) u obtención rápida de ayuda de cualquier objeto (`?`).
* Edición de código `in-line`.
* Introspección de objetos y extracción automática de documentación de los objetos de Python como clases y funciones.
* Buena interacción con el sistema operativo.
* Soporte para ejecutar múltiples procesos en paralelo y en segundo plano que pueden ser ejecutados en clústeres de ordenadores o servicios en la nube como Amazon EC2.

#### IPython Notebook (Jupyter)

Otra forma muy útil es usar los [IPython notebook](http://ipython.org/notebook.html), que es un entorno Web para Python apropiado para realizar computación interactiva. Aunque los *notebooks* utilicen un navegador web como interfaz gráfica éstos se ejecutan localmente (en el mismo ordenador que ejecuta el navegador web).

<div class="alert alert-info">Si bien nació como parte del proyecto IPython, el mismo entorno visual se puede conectar a *"kernels"* de distintos lenguajes. Se puede usar Jupyter con Python, Julia, R, Octave y decenas de lenguajes más.</div>

Podemos crear y editar "celdas" de código Python que podéis editar y volver a ejecutar, se pueden intercalar celdas de texto, fórmulas matemáticas, y hacer que gráficos se muestren incrustados en la misma pantalla. Estos archivos se guardan con extensión *.ipynb*, que pueden exportarse a diversos formatos estáticos como HTML o como código Python puro. (.py)

Los notebooks son muy útiles para la **"programación exploratoria"**, muy frecuente en ciencia e ingeniería

Gran parte del material de esta asignatura estará en formato notebook.

Para ejecutar IPython Notebook, si ejecutamos la consola (`cmd.exe`) y luego `ipython notebook` o `jupyter notebook`, se abrirá el navegador por defecto del ordenador con el entorno IPython notebook. Lo más conveniente es ejecutar la consola (`cmd.exe`), acceder a la carpeta donde esté almacenado el *notebook* a ejecutar (para moverse por las carpeta desde la consola se utiliza el comando cd y el nombre de la carpeta a la que acceder: `cd nombreCarpeta`. Para ir a una carpeta en un nivel superior se utiliza el comando `cd ..`) y ejecutar el comando `ipython notebook` o `jupyter notebook`.

### ¿Qué editor usar?

Python no exige un editor específico y hay muchos modos y maneras de programar. 

#### Spyder

[Spyder](http://code.google.com/p/spyderlib/) es un entorno de desarrollo integrado (IDE) tipo MATLAB para computación científica con Python. Tiene muchas de las ventajas de los IDEs tradicionales como por ejemplo que la edición, ejecución y corrección de errores (debug) de código se realizan en el mismo entorno. Además, se pueden organizar los diferentes códigos en proyectos para mejorar la organización de los mismos.

*Spyder* viene con la distribución Anaconda, por lo que no es necesario instalarlo de forma adicional.

![](images/spyder.png)

#### PyCharm

Otro buen editor es **PyCharm** edición *community*. Es otro ejemplo de IDE tipo MatLab y reúne las ventajas mencionadas para *Spyder*. Sin embargo, este IDE no viene integrado en el entorno Anaconda por lo que se debe descargar e instalar posteriormente. En enlace de descarga es el siguiente: https://www.jetbrains.com/pycharm/
