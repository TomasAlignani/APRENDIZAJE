### Números: enteros, reales y complejos

Python es un lenguaje de muy alto nivel y por lo tanto trae muchos *tipos* de datos incluidos. En esta primera parte vamos a utilizar los tipos de datos numéricos.


Los tipos numéricos básicos son *int* (enteros sin limite), *float* (reales) y *complex* (complejos).

Los números enteros los definimos sin incluir decimales.

```python
1
```

Para comprobar el tipo de dato de un valor (o variable) debemos utilizar la función *type* del siguiente modo.

```python
type(1)
```

Los números reales los definimos con decimales.

```python
1.0
```

```python
type(1.0)
```

```python
1 + 1.4 - 12
```

Los números complejos los definimos estableciendo su parte real y su parte imaginaria (mediante la letra *j*)

```python
3.2 + 12j
```

```python
type(3.2 + 12j)
```

Las operaciones aritméticas básicas son:

* adición: `+`
* sustracción: `-`
* multiplicación: `*`
* división: `/`
* módulo (resto de división): `%`  
* potencia: `**`

Si uno de los datos involucrados en la operación es un número real el resultado es un número real. Si todos los datos son números enteros el resultado es un número entero.

```python
1 + 3
```

```python
type(1 + 3)
```

```python
2.4 - 1
```

```python
type(2.4 - 1)
```

```python
4.6 * 3
```

```python
2**3
```

```python
9/3
```

```python
5/3
```

```python
5.0/3
```

```python
5 % 3
```

Las operaciones entre números complejos son iguales que las anteriores.

```python
(3.2 + 12j) + (1.2 + 2j)
```

```python
(3.2 + 12j) * 2
```

```python
(3.2 + 12j) / 2
```

Las operaciones se pueden agrupar con paréntesis y tienen precedencia estándar.

```python
((21.2 + 4.5)**0.2 / 1.32)
```

Finalmente, los operadores para realizar comparaciones entre datos numéricos (independientemente de su tipo) son:
* `>`: Mayor 
* `<`: Menor 
* `>=`: Mayor o igual 
+ `<=`: Menor o igual
* `==` igualdad


```python
2 > 1
```

```python
2 < 1.0
```

```python
2 > 2
```

```python
2 < 2
```

```python
2 >= 2
```

```python
2 <= 2
```

```python
# igualdad
1.2 == 1.2
```

El resultado de las comparaciones anteriores da como resultado otro tipo de dato básico: los booleanos (*bool*). Este tipo de datos puede tomar dos valores: *True* y *False*.

```python
type(1==1)
```

Los operadores entre datos booleanos son 3:
* Conjunción: para lo que se utiliza la palabra `and`.
* Disyunción: para lo que se utiliza la palabra `or`.
* Negación: para lo que se utiliza la palabra `not`.

```python
True and False
```

```python
True or False
```

```python
not False
```

### Módulos en Python

Un módulo es una porción de un programa en el que se almacenan funcionalidades comunes a una tarea. Para utilizar un módulo en un programa de Python lo primero que se debe hacer es importarlo. Para ello se debe utilizar la instrucción `import`. 

Por ejemplo, un módulo muy útil es el llamado `math` ya que contiene muchas funciones matemáticas estándar. Para importarlo se debe ejecutar la instrucción `import math`.

```python
import math
```

Al importar el módulo de este modo se tienen disponibles todas las funciones asociadas al mismo para poder ser utilizadas en el programa. Para poder utilizar una función se debe poner el nombre del módulo seguido de un punto (`.`) y el nombre de la función a utilizar. Por ejemplo calcular el seno de un número se utiliza la función *sin* del módulo *math*, por lo que se utilizaría ejecutando la instrucción `math.sin(valor)`.

```python
math.sin(7.0)
```

En este módulo también están definidos valores para números especiales como el número pi: `math.pi`

```python
math.pi
```

```python
math.sin(2*math.pi)
```

Algunos ejemplos (más adelante veremos cómo saber todas las funciones contenidas en un módulo) de otras funciones del módulo *math* son:

```python
# Redondeo hacia arriba
math.ceil(5.4)
```

```python
# Truncado
math.trunc(5.8)
```

```python
# Factorial
math.factorial(5)
```

```python
# Raíz cuadrada
math.sqrt(2)
```

```python
# Ejemplo del fallo ya que no existe la raíz cuadrada de -1
math.sqrt(-1) 
```

Una segunda forma de importar un módulo es importar todos su contenido sin tener que poner el nombre del módulo para utilizar sus funciones. Por ejemplo si se desea importar todo el contenido del módulo *math* se ejecutaría la instrucción: `from math import *`

```python
from math import *

print(sin(2 * pi))
```

Esta forma de importar un módulo puede parecer muy buena. Sin embargo, en programas muy grandes que incluyan muchos módulos es una buena idea importar de la primera forma para eliminar la posibilidad de tener colisiones entre funciones de módulos diferentes (una colisión se produciría si dos módulos tienen una función llamada igual).

La tercera alternativa es importar solamente las funciones que se vayan a utilizar. Para ello, siguiendo el patrón de importación anterior podemos especificar las funciones a importar mediante la lista de sus nombres en lugar de utilizar el símbolo `*`.

```python
from math import sin, pi

x = sin(2 * pi)

print(x)
```

### Conociendo los contenidos de un módulo y su documentación

Una pregunta habitual al importar un módulo es: ¿Qué funciones contiene?. Para responderla se puede utilizar la función *dir* una vez importado el módulo.

```python
import math

print(dir(math))
```

Se pueden observar los nombres de todas las funciones contenidas en el módulo *math*.

Además, si se utiliza la función `help` se puede obtener una descripción de cada función. Para ello la función debe tener su descripción (técnicamente se llama docstring) pero la gran mayoría la tienen disponible. 

```python
help(math.log)
```

```python
log(10)
```

```python
log(10, 2)
```

También se puede utilizar la función `help` directamente sobre un módulo. Por ejemplo: `help(math)` 

La lista completa de módulos estándar para Python 2 y Python 3 está disponible en http://docs.python.org/2/library/ y http://docs.python.org/3/library/, respectivamente.

Existe un módulo equivalente para operaciones sobre el dominio complejo. Se llama *cmath*.

```python
import cmath
cmath.sqrt(-1)
```

Los objetos no solo *guardan* valores (atributos) sino que tienen acceso a *métodos*, es decir, traen acciones (funciones) que podemos ejecutar sobre esos valores, a veces requiriendo/permitiendo parámetros adicionales. 

Por ejemplo, en el caso de los *strings* cuando creamos una variable de este tipo lo que estamos creando es un objeto de tipo *string*. La variable contiene el valor que le es asignado y, además, este tipo de dato tiene sus propios **métodos**: pasar a mayúsculas, capitalizar y un largo etcétera.

### Indizado y troceado

Las cadenas son **secuencias**. O sea, conjuntos ordenados que se pueden indizar, recortar, reordenar, etc. 

![](images/index_slicing.png)


Podemos acceder a un carácter determinado espcificando la posición deseada entre corchetes `[posición]`. La indexación en Python comienza en 0.

```python
s = "Hola mundo"
print(s[0])
```

Al igual que se puede realizar con **MatLab** se puede extraer un subconjunto de elementos de una cadena. Para ello se utiliza la sintaxis `[inicio:fin]`, que extrae todos los caracteres comprendidos entre el índice `inicio` y el índice `fin` -1 (el carácter en el índice `fin` no se incluye en el subconjunto extraído.

```python
cadena = "HOLA MUNDO"
print(cadena[0:4])
print(s[5:6])
```

Si omitimos el índice `inicio` o el índice `fin` (o ambos) de `[inicio:fin]`, los índices utilizados por defecto son el comienzo (0) y el final del string, respectivamente.

```python
print(s[:5])
```

```python
print(s[6:])
```

```python
print(s[:])
```

En la sintaxis anterior el tamaño del paso entre `inicio` y `fin` es 1. Es decir, se cogen todos los caracteres de uno en uno. Sin embargo, se puede especificar el tamaño del paso mediante la sintaxis `[inicio:fin:paso]`.

Esta técnica se llama **slicing**. Puedes encontrar más información de esta sintaxis en: http://docs.python.org/release/2.7.3/library/functions.html?highlight=slice#slice

```python
# Todas las letras de una en una
print(s[::1])
```

```python
# Todas las letras de dos en dos
print(s[::2])
```

Se pueden coger los caracteres en orden inverso si el tamaño del paso es negativo (por ejemplo -1).

```python
# Todas las letras den orden inverso
print(cadena[::-1]) 
```

El tipo `str` en python es **inmutable**, lo que quiere decir que, una vez definido un objeto tipo cadena no podemos modificarlo. 

```python
cadena[0] = 'B'
```

Pero si se puede utilizar un string para **crear otro**

```python
# Se concatena la letra B con una subcadena de otro string
'B' + cadena[1:]
```

```python
cadena[:4] + " SUB" + cadena[5:]
```

Se puede cambiar un carácter por otro utilizando la función `replace`.

```python
cadena = cadena.replace('H', 'B')
print(cadena)
```

### Obtención de palabras a partir de una frase

Para obtener las palabras de una frase se utiliza la función `split`. Por defecto se divide por los espacios en blanco pero se puede especificar el símbolo deseado para dividir la cadena de texto en diferentes subcadenas.

```python
a = "hola amigos"

print(a.split())

a = "hola,amigos,como"

palabras = a.split(',')

print(palabras)
```

Este método da como resultado una lista de palabras. Para acceder a cada palabra (o subconjunto de palabras, se puede utilizar la indexación vista anteriormente.

```python
palabras[0]
```

```python
palabras[:2]
```

### Interpolación

Se puede crear un string a partir de una "plantilla" con un formato predeterminado. La forma más poderosa es a través del método [`format`](https://docs.python.org/3.4/library/string.html#format-examples)

```python
"{} es {}".format('Messi', 'crack')   # por posición, implicito
```

```python
"{1} es {0}".format('Messi', 'crack')     # por posición, explícito
```

```python
"{saludo} {planeta}".format(saludo='Hola', planeta='Mundo')    # por nombre de argumentos
```

```python
"La parte real es {numero.real:.5f} y la imaginaria es {numero.imag}  {numero}".format(numero=x)
```

## Casting de tipos

Python es un lenguaje de tipado dinámico como se ha mencionado anteriormente. Es conveniente recalcar que su **tipado es fuerte**. Quiere decir que no intenta adivinar los tipos de los datos y exige ser explícito en ellos.

Se puede cambiar el tipo de dato escribiendo el tipo de dato al que queremos cambiar (int, float, complex, bool, str) y entre paréntesis el dato al que se desea cambiar su tipo.

```python
x = 1
type(x)
```

```python
y = float(x)
type(y)
```

```python
y = complex(x)
type(y)
```

```python
y = str(x)
type(y)
```

Se pueden realizar operaciones entre datos que hayan sido cambiados de tipo mediante el correspondiente cast.

```python
"2" + "2"
```

```python
int("2") + int("2")
```

```python
float('2.34545') ** 2
```

```python
x = 1.5

print(x, type(x))
```

```python
x = int(x)

print(x, type(x))
```

```python
z = complex(x)

print(z, type(z))
```

Las variables con números complejos no pueden ser cambiadas (cast) a tipos de datos enteros o reales. Para ello se debe utilizar `z.real` o `z.imag` para extraer cada parte del número complejo y hacer su cast correspondiente.

```python
x = float(z)
```

```python
float(z.real)
```

```python
float(z.imag)
```
