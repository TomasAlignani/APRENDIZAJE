### Ejercicio de aproximación de $\pi$
Aproximación del número $\pi$ mediante números aleatorios. Generar un conjunto de $n$ puntos aleatorios de dos dimensiones en el cuadrado $[0,1]$. Para cada pareja $(𝑥_1,𝑥_2)$, contamos cuántas parejas satisfacen que $𝑥_1^2+𝑥_2^2\leq 1$. Podemos asumir que estos puntos caen dentro de un círculo de radio $1$ centrado en el punto $(0,0)$. El área de todo el círculo es $\pi r^2 = \pi$ (puesto que $r=1$). Por tanto, como solo trabajamos con un cuarto de círculo, la aproximación de $\pi$ será 

$$
4\dfrac{\text{\#Puntos dentro del círculo}}{n}
$$

Evidentemente, a mayor número de puntos aleatorios, mejor aproximación obtendremos de $\pi$.
