# NumPy: Aplicación en Análisis de Componentes Principales

El Análisis de Componentes Principales (PCA en inglés) es una técnica de reducción de la dimensión de un problema. En resumen, cuando los objetos del problema están representados por un número muy grande de variables (dimensiones), el objetivo de PCA es reducirlas de tal manera que perdamos la menor cantidad de información posible. Por ejemplo, si en nuestro problema los objetos están representados por 1000 dimensiones (1000 variables diferentes), PCA trata de reducirlas (por ejemplo a 10 variables) que sirvan para distinguirlos, clasificarlos, etc. de la misma manera, pero reduciendo el coste computacional o de almacenamiento. Además, PCA puede ser muy útil para la visualización de datos, ya que podríamos reducir cualquier objeto multidimensional a 2 o 3 dimensiones, que pueden ser fácilmente graficadas. 

Aunque para entender perfectamente PCA son necesario ciertos conocimientos de álgebra lineal, para nosotros es suficiente saber que PCA  trata de encontrar las direcciones de máxima variabilidad de los datos (componentes principales). Para ello, se analiza la correlación entre las distintas dimensiones del problema, construyendo un nuevo sistema de referencia a partir de los vectores propios de la matriz de correlación.

### Una forma sencilla de reconocer automáticamente nuevas imágenes

Además de las 3000 imágenes que forman el conjunto de entrenamiento, disponemos de 2000 nuevas imágenes a las que vamos a llamar conjunto de test. Estas imágenes son las que vamos a utilizar para determinar si nuestro clasificador automático de dígitos funciona bien o funciona mal. Pero, ¿cómo podemos implementar un reconocedor automático de manera muy sencilla? Lee los siguientes pasos:
1. Leer una imagen del conjunto de test
2. Medir la distancia entre dicha imagen de test y cada una de las 3000 imágenes de entrenamiento. Con esto obtenemos un valor numérico que nos dice lo distantes que son dichas imágenes
3. Buscar el valor de distancia más pequeño. Clasificaremos la imagen de test con el dígito asociado a la imagen de entrenamiento de menor distancia. Es decir, si resulta que la imagen más cercana a la imagen de test es la imagen 1547 (que representa al dígito 5), entonces clasificaremos la imagen de test como dígito 5. 

Si repetimos este proceso para cada una de las 2000 imágenes de test, podemos obtener un porcentaje de muestras bien clasificadas y, por tanto, una métrica de lo bueno o malo que es mi clasificador.
